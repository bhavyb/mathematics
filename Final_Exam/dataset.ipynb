```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 5000

student_id = [f"S{1000+i}" for i in range(n)]
age = np.random.randint(15, 19, n)
hours_studied = np.round(np.random.normal(5, 2, n).clip(0, 12), 1)
math_score = np.round(np.random.normal(60, 15, n).clip(0, 100), 1)
science_score = np.round(np.random.normal(58, 14, n).clip(0, 100), 1)
english_score = np.round(np.random.normal(65, 12, n).clip(0, 100), 1)

# Pass_Fail logic based on average score + hours studied influence
avg_score = (math_score + science_score + english_score) / 3
pass_prob = 1 / (1 + np.exp(-(avg_score - 50)/10 + (hours_studied-5)/10))
pass_fail = (np.random.rand(n) < pass_prob).astype(int)

df = pd.DataFrame({
    "Student_ID": student_id,
    "Age": age,
    "Math_Score": math_score,
    "Science_Score": science_score,
    "English_Score": english_score,
    "Hours_Studied": hours_studied,
    "Pass_Fail": pass_fail
})

df.to_csv("students_scores.csv", index=False)
print(df.head())
```
